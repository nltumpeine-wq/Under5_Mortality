country-level entities and annual observations from 1960 to 2025.

- 217 countries
- 66 years
- 14,322 country-year observations

However, the target contains missing values.

The year 2025 has 100% missing target values across the
217 countries and therefore cannot be used as a target year.

Recent years from 2010–2024 show approximately 9.68% missing
target observations per year.

The remaining missingness will be investigated by country

#### **Trying to investigate patterns in missing data in basic sanitation column under mortality_sanitation dataset**

What happened to the 259 missing sanitation values?
This is actually a great data-science observation.
Before merging:
Sanitation missing = 259
After merging:
Sanitation missing = 187
Why did 72 disappear?
Because those 72 missing sanitation rows belonged to entities/year combinations that didn't appear in mortality_target, so the inner merge excluded them.
Remember:
inner merge keeps only Country Code + Year combinations that exist on both sides.
So:
259 sanitation missing    ↓
only those belonging to our mortality population
187 remaining

## Model Evaluation and Selection

Eight regression approaches were evaluated for predicting
under-five mortality using four selected health and socioeconomic
indicators.

The models included a baseline predictor, Linear Regression,
Decision Trees, Random Forests, and Artificial Neural Networks.

### Results

The tuned Random Forest achieved the strongest performance
among the evaluated models, with:

- MAE: 8.73
- RMSE: 18.83
- R²: 0.656

The model performed better than the baseline and the other
models in the reported comparison.

The results suggest that the selected indicators contain useful
predictive information about under-five mortality. However,
prediction errors remain substantial for some observations,
particularly those with high mortality values.

### Limitations

The dataset has incomplete coverage across countries and years.
Consequently, the evaluation results may not represent every
country equally.

The observed relationships do not establish causation.
Furthermore, model performance alone does not demonstrate
that any individual health indicator causes mortality to increase
or decrease.

The tuned Random Forest was selected as the preferred model
